```python
import numpy as np

def load_optimize_data(csv_path="optimizeData_02232026.csv", has_header=True):
    """
    Load optimize CSV data into a NumPy array.
    """
    skip = 1 if has_header else 0
    data = np.genfromtxt(csv_path, delimiter=",", skip_header=skip)
    return np.atleast_2d(data)
```

```python
arr = load_optimize_data("optimizeData_02232026.csv", has_header=False)
print(arr.shape)
fitness = arr[:, -1]
gom = np.abs(arr[:, -2])
```

```python
import matplotlib.pyplot as plt
plt.plot(fitness)
```
